```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df =pd.read_csv("fifa_data.csv")
print(df)
```

1.Which country has the most number of players (score :1)

```python
country = df["Nationality"].value_counts().index[0]
print(country)
```

England has the highest number of players in the data set.

This shows that England produces more football players compared to other countries and football is very popular there.

That is why there are many players are developed in this country.

2.Plot a bar chart of 5 top countries with the most number of players. (score :1)

```python
import matplotlib.pyplot as plt

countries = df["Nationality"].value_counts().head(5)
countries.plot(kind = "bar",color="teal")
plt.xlabel("Country")
plt.ylabel("Number of players")
plt.title("Top 5 countries with most number of players")
plt.show()
```

 The bar chart shows the Top 5 countries with the most FIFA players.
 
 It shows the football is more popular in these regions.
 
 These countries produce more players compared to others.

3.Which player has the highest salary? (score :1)

```python

df["Wage"]=df["Wage"].str.replace("€","").str.replace("K","").astype(float)
highest_salary = df.loc[df["Wage"].idxmax()]


print(highest_salary["Name"])
print(highest_salary["Wage"])
```

The highest - paid player is Lionel Messi.

This shows that top players earn much more than other players due to their skill,performance,and popularity.

4.Plot a histogram to get the salary range of the players. (score :1)

```python
import matplotlib.pyplot as plt

plt.hist(df["Wage"],color = "purple",bins = 20)
plt.xlabel("Wage")
plt.ylabel("Players")
plt.title("Wage Distribution")
plt.show()
```

The histogram shows that most players earn lower wages,while only a few players earn very high salaries.

This means high salaries are rare,and most players fall into the lower salary range.

5.Who is the tallest player in the fifa? (score :1)

```python
tallest_player = df.sort_values(by = "Height",ascending = False).iloc[0]
print(tallest_player["Name"])
print(tallest_player["Height"])
```

The tallest player is T.Holy with a height of 6'9.

This shows that very tall players are rare in football .

Tall players have advantages like strong ability , and are especially useful in positions like goal keeper and defender.

7.Which foot is most preferred by the players?Draw a bar chart for preferred foot (score :1)

```python
foot_count = df["Preferred Foot"].value_counts()
print(foot_count)
```

```python
import matplotlib.pyplot as plt

foot_count.plot(kind = "bar",color = "orange")
plt.xlabel("Foot")
plt.ylabel("Number of Players")
plt.title("Preferred Foot Players")
plt.show()
```

Most players prefer the right foot,showing that right-footed players are common in football.

Because most people naturally use their right foot more.

Teams also tend to have more right-footed players,while left-footed players are fewer and sometimes considered special or valuable in certain positions.
